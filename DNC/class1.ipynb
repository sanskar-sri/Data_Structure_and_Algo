o/p:- [10,15,20,25,30,40]

```python
def merge_sorting(arr1,size1,index1,arr2,size2,index2,ans):
    
    ##comparison 
    while(index1<size1 and index2<size2):
        if(arr1[index1]<arr2[index2]):
            ans.append(arr1[index1])
            index1=index1+1
        else:
            ans.append(arr2[index2])
            index2=index2+1
            
    ##copying
    while(index1<size1):
        ans.append(arr1[index1])
        index1=index1+1
    while(index2<size2):
        ans.append(arr2[index2])
        index2=index2+1
        
    


arr1=[10,20,30,40]
arr2=[5,15,25]
index1=0
index2=0
size1=len(arr1)
size2=len(arr2)
ans=[]
merge_sorting(arr1,size1,index1,arr2,size2,index2,ans)
print(ans)
```

```python
def merge(arr,size,start,end,mid):
    left_len=mid-start+1
    right_len=end-mid
    
    ##creating temp list with 0 placeholder 
    arr1=[0]*left_len
    arr2=[0]*right_len
    
    ##now copying the element from start to mid
    for i in range(left_len):
        arr1[i]=arr[start+i]
    ##copying element from mid+1 to end
    for j in range(right_len):
        arr2[j]=arr[mid+1+j]
        
    
    i=0
    j=0
    k=start
    
    ##comparison
    while(i<len(arr1) and j< len(arr2)):
        if(arr1[i]<arr2[j]):
            arr[k]=arr1[i]
            i+=1
        else:
            arr[k]=arr2[j]
            j+=1
        k=k+1
     
    ##copying
    while(i<len(arr1)):
        arr[k]=arr1[i]
        i+=1
        k=k+1
        
    while(j<len(arr2)):
        arr[k]=arr2[j]
        j+=1
        k=k+1
                    

def merge_sort(arr,size,start,end):
    
    ##base conditon
    if(start>=end):
        return
    
    ##mid wala index nikala
    mid=(start+end)//2
    
    ##left part ko recursion se krvana  hai
    merge_sort(arr,size,start,mid)
    
    ##right part ko recursion se krvana hai
    merge_sort(arr,size,mid+1,end)
    
    ##recusion ne jo sorted array diye usko merge kardo
    merge(arr,size,start,end,mid)
    

arr=[5,7,2,4,1,3,9,8]
size=len(arr)
start=0
end=size-1

merge_sort(arr,size,start,end)
print(arr)
```

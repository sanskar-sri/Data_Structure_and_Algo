MERGE 2 SORTED ARRAY

Q) 2 SORTED ARRAY 
    arr1=[10,20,30,40]
    arr2=[15,25]
o/p:- [10,15,20,25,30,40]

```python
def merge_sorting(arr1,size1,index1,arr2,size2,index2,ans):
    
    ##comparison 
    while(index1<size1 and index2<size2):
        if(arr1[index1]<arr2[index2]):
            ans.append(arr1[index1])
            index1=index1+1
        else:
            ans.append(arr2[index2])
            index2=index2+1
            
    ##copying
    while(index1<size1):
        ans.append(arr1[index1])
        index1=index1+1
    while(index2<size2):
        ans.append(arr2[index2])
        index2=index2+1
        
    


arr1=[10,20,30,40]
arr2=[5,15,25]
index1=0
index2=0
size1=len(arr1)
size2=len(arr2)
ans=[]
merge_sorting(arr1,size1,index1,arr2,size2,index2,ans)
print(ans)
```

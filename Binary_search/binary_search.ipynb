Binary search without recursion

T.C->O(LOG N)              S.C->O(1)

```python
def binary_search(arr,target,size):
    start=0
    end=size-1
    
    while(start<=end):
        mid=(start+end)//2
        if(arr[mid]==target):
            return True
        if(target<arr[mid]):
            end=mid-1
        else:
            start=mid+1
            
    return False
        

arr=[10,20,30,40]
target=201
index=0
size=len(arr)
ans=binary_search(arr,target,size)
print(ans)
```

Binary search with recursion

T.C->O(LOG N)   S.C->O(LOG N)

```python
def binary_search(arr,target,size,s,e):
    
    ##base condition
    if(s>e):
        return False
    
    mid=(s+e)//2
    if(arr[mid]==target):
        return True
    if(target<arr[mid]):
        return binary_search(arr,target,size,s,mid-1)
    else:
        return binary_search(arr,target,size,mid+1,e)
        

arr=[10,20,30,40]
target=20
size=len(arr)
s=0
e=size-1
ans=binary_search(arr,target,size,s,e)
print(ans)
```

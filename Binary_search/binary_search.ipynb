##binary search

```python
def binary_search(arr,target,size):
    start=0
    end=size-1
    
    while(start<=end):
        mid=(start+end)//2
        if(arr[mid]==target):
            return True
        if(target<arr[mid]):
            end=mid-1
        else:
            start=mid+1
            
    return False
        

arr=[10,20,30,40]
target=201
index=0
size=len(arr)
ans=binary_search(arr,target,size)
print(ans)
```

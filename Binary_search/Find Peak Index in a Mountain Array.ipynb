Find Peak Index in a Mountain Array

Has at least 3 elements.

Contains a peak element where:

arr[0] < arr[1] < ... < arr[i]

arr[i] > arr[i + 1] > ... > arr[arr.length - 1]

Your task is to find the index of the peak element in the given mountain array arr.

Input: arr = [0, 2, 5, 3, 1]
Output: 2

Constraints
1 <= nums.length <= 100
0 <= nums[i] <= 1000

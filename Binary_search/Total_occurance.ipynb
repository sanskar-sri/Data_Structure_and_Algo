Finding Total Occurrences of an Element in a Sorted Array

To find the total number of occurrences, compute the first and last occurrences, and subtract:

Input: nums = [5,7,7,8,8,10], target = 8
Output: 2

O(log n) time and O(1)

```python
def first_occ(nums,target,start,end):
    
    ans1=-1
    while(start<=end):
        mid=(start+end)//2
        ##store and compute
        if(nums[mid]==target):
            ans1=mid
            end=mid-1
        elif(target<nums[mid]):
            end=mid-1
        else:
            start=mid+1
            
    return ans1
    
def last_occ(nums,target,start,end):
    
    ans2=-1
    while(start<=end):
        mid=(start+end)//2
        ##store and compute
        if(nums[mid]==target):
            ans2=mid
            start=mid+1
        elif(target<nums[mid]):
            end=mid-1
        else:
            start=mid+1
    return ans2
    

def total_occ(nums,target,size):
    start=0
    end=size-1
    ans2=-1
    a1=first_occ(nums,target,start,end)
    a2=last_occ(nums,target,start,end)

    return (a2-a1)+1
    
nums = [5,7,7,8,8,10]
target = 8
size=len(nums)
final_occ=total_occ(nums,target,size)
print(final_occ)
```

WITH RECURSION

O(log n) time and O(log n)

```python
def first_occ(nums,target,start,end,ans1):
    ##base condition
    if(start>end):
        return ans1
    
    ##compute
    mid=(start+end)//2
    if(nums[mid]==target):
        ans1=mid
        return first_occ(nums,target,start,mid-1,ans1)
    elif(target<nums[mid]):
        return first_occ(nums,target,start,mid-1,ans1)
    else:
        return first_occ(nums,target,mid+1,end,ans1)


def last_occ(nums,target,start,end,ans2):
    
     ##base condition
    if(start>end):
        return ans2
    
    mid=(start+end)//2
    if(nums[mid]==target):
        ans2=mid
        return last_occ(nums,target,mid+1,end,ans2)
    elif(target<nums[mid]):
        return last_occ(nums,target,start,mid-1,ans2)
    else:
        return last_occ(nums,target,mid+1,end,ans2)
    
    

def total_occ(nums,target,size):
    start=0
    end=size-1
    ans=[]
   
    a1=first_occ(nums,target,start,end,-1)
    a2=last_occ(nums,target,start,end,-1)
    
    ans.append(a1)
    ans.append(a2)
    return ans
    
nums = [5,7,7,8,8,10]
target = 8
size=len(nums)
final_occ=total_occ(nums,target,size)
print(final_occ)
```
